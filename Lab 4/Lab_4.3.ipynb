Task Description#1
●	Zero-shot: Prompt AI to write a function that checks whether a given year is a leap year.
Expected Output#1
●	AI-generated function with no examples provided.

Prompt: Create a function that checks whether a given year is a leap year by taking input from console.



```python
def is_leap_year():
  """
  Checks if a given year is a leap year based on user input from the console.
  """
  try:
    year = int(input("Enter a year: "))
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
      print(f"{year} is a leap year.")
    else:
      print(f"{year} is not a leap year.")
  except ValueError:
    print("Invalid input. Please enter a valid year (a whole number).")

# Call the function to test it
is_leap_year()
```

Task Description#2
●	One-shot: Give one input-output example to guide AI in writing a function that converts centimeters to inches.
Expected Output#2
●	Function with correct conversion logic

Prompt: Create a function that converts centimeters to inches.
Ex: enter the centimeters:  2.5
The 2.5cm is 1 inches


```python
def cm_to_inches():
  """
  Converts centimeters to inches based on user input.
  """
  try:
    cm = float(input("enter the centimeters: "))
    inches = cm / 2.54  # 1 inch = 2.54 cm
    print(f"The {cm}cm is {inches} inches")
  except ValueError:
    print("Invalid input. Please enter a valid number for centimeters.")

# Call the function to test it
cm_to_inches()
```

Prompt: Create a function that formats full names as “Last, First” by taking input form the console .
Ex: enter the first name: suchith

Task Description#4
●	Compare zero-shot and few-shot prompts for writing a function that counts the number of vowels in a string.
Expected Output#4
●	Functional output and comparative reflection

Zero-shot Prompt: create a function that counts the number of vowels in a string by taking input from the console.


```python
def count_vowels():
  """
  Counts the number of vowels in a string based on user input.
  """
  input_string = input("Enter a string: ")
  vowels = "aeiouAEIOU"
  vowel_count = 0
  for char in input_string:
    if char in vowels:
      vowel_count += 1
  print(f"The number of vowels in '{input_string}' is: {vowel_count}")

# Call the function to test it
count_vowels()
```

```python
def count_vowels_few_shot():
  """
  Counts the number of vowels in a string based on user input,
  following the format of the few-shot examples.
  """
  input_string = input("Enter the string: ")
  vowels = "aeiouAEIOU"
  vowel_count = 0
  for char in input_string:
    if char in vowels:
      vowel_count += 1
  print(f"The no. of vowels in the string: {vowel_count}")

# Call the function to test it
count_vowels_few_shot()
```

Task Description#5
●	Use few-shot prompting to generate a function that reads a .txt file and returns the number of lines.
Expected Output#5
●	Working file-processing function with AI-guided logic

 Prompt: Create generate a function that reads a .txt file and returns the number of lines by taking the file name in the console.
Ex: enter the file name: skk.text
The no. of lines in the file are 4 lines
Enter the file name: hi.text
The no.of lines in the file are 2 lines


```python
def count_lines_in_file():
  """
  Reads a text file and returns the number of lines based on user input filename.
  """
  file_name = input("Enter the file name: ")
  try:
    with open(file_name, 'r') as f:
      lines = f.readlines()
      num_lines = len(lines)
      print(f"The no. of lines in the file are {num_lines} lines")
  except FileNotFoundError:
    print(f"Error: File '{file_name}' not found.")
  except Exception as e:
    print(f"An error occurred: {e}")

# Call the function to test it
count_lines_in_file()
```
